```python
from Bio import Entrez
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import time

# Always provide your email when using Entrez
Entrez.email = "user@example.com"

def get_pubmed_counts(term, start_year, end_year):
    counts = []
    for year in range(start_year, end_year + 1):
        # Search term structured as described in the methods
        search_query = f"{term} AND ({year}[pdat])"
        try:
            # retmax=0 because we only need the Count metadata
            handle = Entrez.esearch(db="pubmed", term=search_query, retmax=0)
            record = Entrez.read(handle)
            handle.close()
            counts.append(int(record["Count"]))
        except Exception as e:
            print(f"Error fetching data for {year}: {e}")
            counts.append(0)
        # Respect NCBI rate limits (3 requests per second max without API key)
        time.sleep(0.35)
    return counts
```

```python
start_year = 2000
end_year = 2025
years = list(range(start_year, end_year + 1))

print("Fetching neuroscience publications...")
neuroscience_counts = get_pubmed_counts("neuroscience", start_year, end_year)

print("Fetching neuro oncology publications...")
neuro_onc_counts = get_pubmed_counts("neuro oncology", start_year, end_year)

print("Fetching neurology publications...")
neurology_counts = get_pubmed_counts("neurology", start_year, end_year)

# Create DataFrame
df_pubs = pd.DataFrame({
    'Year': years,
    'Neuroscience': neuroscience_counts,
    'Neuro-oncology': neuro_onc_counts,
    'Neurology': neurology_counts
})

# Calculate proportion
df_pubs['Neuro_Onc_to_Neurology_Ratio'] = df_pubs['Neuro-oncology'] / df_pubs['Neurology']

display(df_pubs.head())
```

```python
# Set seaborn style
sns.set_theme(style="whitegrid")

# Plot 1: Neuroscience vs Neuro-oncology
plt.figure(figsize=(10, 6))
sns.lineplot(data=df_pubs, x='Year', y='Neuroscience', label='Neuroscience', marker='o')
sns.lineplot(data=df_pubs, x='Year', y='Neuro-oncology', label='Neuro-oncology', marker='s')
plt.title('Neuroscience vs. Neuro-oncology Publications (2000-2025)')
plt.ylabel('Number of Publications')
plt.xlabel('Year')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Plot 2: Proportion of Neuro-oncology to Neurology
plt.figure(figsize=(10, 6))
sns.lineplot(data=df_pubs, x='Year', y='Neuro_Onc_to_Neurology_Ratio', color='purple', marker='^')
plt.title('Proportion of Neuro-oncology to Neurology Publications (2000-2025)')
plt.ylabel('Ratio (Neuro-oncology / Neurology)')
plt.xlabel('Year')
plt.tight_layout()
plt.show()
```
